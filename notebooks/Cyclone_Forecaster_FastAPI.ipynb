```python
import nest_asyncio
import uvicorn
from fastapi import FastAPI
from fastapi.middleware.cors import CORSMiddleware

nest_asyncio.apply()

app = FastAPI(
    title="Cyclone Impact & Infrastructure Vulnerability API",
    description="AI-powered cyclone risk and infrastructure exposure backend",
    version="1.0.0"
)

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"]
)

print("FastAPI app created successfully.")
```

```python
@app.get("/")
def root():
    return {
        "project": "Cyclone Impact & Infrastructure Vulnerability Forecaster",
        "status": "online",
        "message": "Cyclone risk intelligence API is running"
    }

@app.get("/health")
def health():
    return {
        "status": "healthy"
    }

print("Endpoints created:")
print("GET /")
print("GET /health")
```

```python
forecast_data = {
    "risk": {
        "prediction_locations": 2560,
        "high_risk_locations": 49,
        "moderate_risk_locations": 574,
        "low_risk_locations": 1937,
        "mean_flood_probability": 0.1635,
        "maximum_flood_probability": 0.5267,
        "high_risk_probability_min": 0.3533,
        "high_risk_probability_max": 0.5267
    },

    "infrastructure": {
        "total_roads": 87306,
        "roads_near_high_risk": 5519,
        "road_exposure_percent": 6.32,
        "total_power_infrastructure": 39168,
        "power_near_high_risk": 553,
        "power_exposure_percent": 1.41,
        "power_grid_near_high_risk": 552,
        "power_plants_near_high_risk": 1,
        "exposure_radius_km": 5
    }
}

print("Forecast data loaded.")
```

```python
advisory = """
1. RISK OVERVIEW
- Prediction locations evaluated: 2,560
- Risk classifications: 49 high-risk, 574 moderate-risk, 1,937 low-risk locations
- RF flood-risk scores: Mean 0.1635, Maximum 0.5267
- High-risk classification threshold: 0.35

2. INFRASTRUCTURE EXPOSURE
- Roads: 5,519 out of 87,306 total road features (6.32%) are located near high-risk locations.
- Power Infrastructure: 553 out of 39,168 total power infrastructure features (1.41%) are located near high-risk locations, comprising 552 power-grid features and 1 power plant.

3. PRIORITY ACTIONS
- Deploy inspection and monitoring teams to focus on infrastructure near the 49 high-risk locations.
- Prioritize contingency planning for the identified power infrastructure.
- Establish traffic management and rerouting plans for vulnerable road segments.

4. LIMITATIONS
- These are model-derived risk and spatial-proximity results.
- They do not establish that infrastructure was actually damaged or flooded.
"""

print("Advisory loaded:", len(advisory), "characters")
```

```python
@app.get("/forecast")
def get_forecast():
    return {
        "status": "success",
        "forecast": forecast_data,
        "advisory": advisory
    }

print("GET /forecast endpoint created.")
```

```python
import threading

def run_api():
    uvicorn.run(
        app,
        host="0.0.0.0",
        port=8000
    )

api_thread = threading.Thread(
    target=run_api,
    daemon=True
)

api_thread.start()

print("FastAPI server started on port 8000.")
```
